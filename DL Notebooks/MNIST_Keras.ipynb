# MNIST - Multi - Classification using Keras

```python
import tensorflow
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import  Dense,Flatten
```

```python
(X_train,y_train),(X_test,y_test) = keras.datasets.mnist.load_data()
```

```python
X_train.shape
```

```python
X_train[0]
```

```python
# pixel value to image 
import matplotlib.pyplot as plt
plt.imshow(X_train[2])
```

```python
# Scaling the X_train
X_train = X_train/255
X_test = X_test/255
```

```python
X_train
```

```python
# Building the Artificial Neural Network
model = Sequential()

model.add(Flatten(input_shape=(28,28))) # Flatten in Deep Learning Convert the 2D arrays into 1-D so here it becomes we want 784 input to feed into nn
model.add(Dense(128,activation='relu'))
model.add(Dense(10 , activation='softmax'))
```

```python
model.summary()
```

```python
model.compile(loss = 'sparse_categorical_crossentropy',optimizer='Adam')
```

```python
model.fit(X_train,y_train,epochs=10,validation_split=0.2)
```

```python
y_prob = model.predict(X_test)
```

```python
y_pred = y_prob.argmax(axis=1)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test , y_pred)
```
