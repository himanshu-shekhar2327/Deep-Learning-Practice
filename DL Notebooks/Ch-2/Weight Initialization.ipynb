# Weight Initialization Procedure in Dl .

# Case 1 : Zero Weight initialization 

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Creating Dataset
from sklearn.datasets import make_classification
import pandas as pd

X, y = make_classification(
    n_samples=120,
    n_features=2,
    n_informative=2,
    n_redundant=0,
    n_classes=2,
    random_state=42
)

df = pd.DataFrame(X, columns=["X", "Y"])
df["Class"] = y

print(df.head())
print(df.shape)
```

```python
plt.scatter(df['X'] , df['Y'] , c = df['Class'])
```

```python
import tensorflow
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense
```

```python
model = Sequential()

# model.add(Dense(10,activation='relu',input_dim=2))
model.add(Dense(10,activation='tanh',input_dim=2))
model.add(Dense(1,activation='sigmoid'))

model.summary()
```

```python
# Set parameters to 0
model.get_weights()
```

```python
initial_weights = model.get_weights()
```

```python
initial_weights[0] = np.zeros(model.get_weights()[0].shape)
initial_weights[1] = np.zeros(model.get_weights()[1].shape)
initial_weights[2] = np.zeros(model.get_weights()[2].shape)
initial_weights[3] = np.zeros(model.get_weights()[3].shape)
```

```python
model.set_weights(initial_weights)
```

```python
model.get_weights()
```

```python
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
history = model.fit(X,y,epochs=100,validation_split=0.2,verbose=0)
```

```python
model.get_weights()
```

```python
from mlxtend.plotting import plot_decision_regions
plot_decision_regions(X,y.astype('int'), clf=model, legend=2)
```

# Using the Sigmoid as a activation Funtion in Hidden Layer

```python
model1 = Sequential()

model1.add(Dense(2,activation='sigmoid',input_dim=2))
model1.add(Dense(1,activation='sigmoid'))

model1.summary()
```

```python
# Set parameters to 0
model1.get_weights()  # THese are the weights given by keras
```

```python
initial_weights1 = model1.get_weights()
```

```python
initial_weights1[0] = np.zeros(model1.get_weights()[0].shape)
initial_weights1[1] = np.zeros(model1.get_weights()[1].shape)
initial_weights1[2] = np.zeros(model1.get_weights()[2].shape)
initial_weights1[3] = np.zeros(model1.get_weights()[3].shape)
```

```python
model1.set_weights(initial_weights1)
```

```python
model1.get_weights()
```

```python
model1.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
history = model1.fit(X,y,epochs=100,validation_split=0.2)
```

```python
model1.get_weights()
```

### Here we can see that the same weight from one input goes to two different neurons . [-0.49216163, -0.49216163]  this weight is from input 1

### [ 0.05185208,  0.05185208] these weights are from input 2 goes to 2 different neurons


```python
from mlxtend.plotting import plot_decision_regions
plot_decision_regions(X,y.astype('int'), clf=model1, legend=2)
```

# Conclusion 
As we see from the above that same weight flows to different neurons . It means no matter how many nodes are there while u are using sigmoid in hiddden layer but still the behave like a  single neuron and acts as linear model. Behaves as Perceptron

# Case 2 : W != 0 , b != 0 i.e (w = 0.5 ,b = 0.5)

```python
model = Sequential()

model.add(Dense(2,activation='relu',input_dim=2))
model.add(Dense(1,activation='sigmoid'))

model.summary()
```

```python
# Set parameters to 0
model.get_weights()
```

```python
initial_weights = model.get_weights()
```

```python
initial_weights[0] = np.ones(model.get_weights()[0].shape)*0.5
initial_weights[1] = np.ones(model.get_weights()[1].shape)*0.5
initial_weights[2] = np.ones(model.get_weights()[2].shape)*0.5
initial_weights[3] = np.ones(model.get_weights()[3].shape)*0.5
```

```python
model.set_weights(initial_weights)
```

```python
model.get_weights()
```

```python
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
history = model.fit(X,y,epochs=100,validation_split=0.2,verbose=0)
```

```python
model.get_weights()
```

```python
from mlxtend.plotting import plot_decision_regions
plot_decision_regions(X,y.astype('int'), clf=model, legend=2)
```

## So we can see that we are getting the same result as in the case 1. 
## SO We can conclude that we cannot initialize our weight with zero and not with constant value
